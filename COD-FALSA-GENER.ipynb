```python
import pandas as pd
```

```python


# Ruta del archivo
ruta_excel = r'C:\MICHAEL-CALIDAD-ENERGIA\PROYECTOS\IDENTIFICACION DE LIBRES CON FALSA GENERACION\CLIENTES-LIBRES.xlsx'

# Cargar cada hoja en un DataFrame
df_centro_control = pd.read_excel(ruta_excel, sheet_name='DATA-CENTRO-CONTROL')
df_gemas = pd.read_excel(ruta_excel, sheet_name='DATA-GEMAS')
df_libres = pd.read_excel(ruta_excel, sheet_name='DATA-LIBRES')



# Asegurar formato datetime
df_centro_control['FechaApertura'] = pd.to_datetime(df_centro_control['FechaApertura'], format='%d/%m/%Y %H:%M:%S', errors='coerce')
df_centro_control['HoraApertura'] = pd.to_datetime(df_centro_control['HoraApertura'], format='%d/%m/%Y %H:%M:%S', errors='coerce')

# Extraer fecha y hora por separado
fechas1 = df_centro_control['FechaApertura'].dt.date
horas1 = df_centro_control['HoraApertura'].dt.time



df_centro_control['FechaCierre'] = pd.to_datetime(df_centro_control['FechaCierre'], format='%d/%m/%Y %H:%M:%S', errors='coerce')
df_centro_control['HoraCierre'] = pd.to_datetime(df_centro_control['HoraCierre'], format='%d/%m/%Y %H:%M:%S', errors='coerce')

# Extraer fecha y hora por separado
fechas2 = df_centro_control['FechaCierre'].dt.date
horas2 = df_centro_control['HoraCierre'].dt.time


# Función segura para combinar fecha y hora
def combinar_seguro(fecha, hora):
    if pd.isna(fecha) or pd.isna(hora):
        return pd.NaT
    return pd.Timestamp.combine(fecha, hora)

# Aplicar combinación segura
df_centro_control['Fecha Inicio'] = [
    combinar_seguro(f, h) for f, h in zip(fechas1, horas1)
]

df_centro_control['Fecha Fin'] = [
    combinar_seguro(f, h) for f, h in zip(fechas2, horas2)
]
# Mostrar resultado
display(df_centro_control)

# Mostrar las primeras filas de cada DataFrame (opcional)
display(df_centro_control)
display(df_gemas)
display(df_libres)
```

```python
# Merge tipo LEFT para conservar todos los clientes libres
df_merged = pd.merge(df_libres, df_gemas, on='COD_DOCUMENTO', how='left')

# Crear columna de observación
df_merged['OBSERVACION'] = df_merged['Causa'].apply(
    lambda x: 'POSIBLE FALSA GENERACION' if pd.isna(x) or str(x).strip() == '' else ''
)


df_merged['Fecha Inicio_y'] = pd.to_datetime(df_merged['Fecha Inicio_y'], format='%d/%m/%Y %H:%M:%S', errors='coerce')
df_merged['Fecha Fin_y'] = pd.to_datetime(df_merged['Fecha Fin_y'], format='%d/%m/%Y %H:%M:%S', errors='coerce')

# Mostrar resultado
#print(df_merged[['COD_DOCUMENTO', 'Causa', 'OBSERVACION']].head())
display(df_merged)
```

```python
# Asegurar que las columnas de fecha estén en formato datetime
df_centro_control['Fecha Inicio'] = pd.to_datetime(df_centro_control['Fecha Inicio'], format='%d/%m/%Y %H:%M:%S', errors='coerce')
df_centro_control['Fecha Fin'] = pd.to_datetime(df_centro_control['Fecha Fin'],format='%d/%m/%Y %H:%M:%S', errors='coerce')

# Agrupar por COD_DOCUMENTO y obtener máximos
df_agrupado = df_centro_control.groupby('COD_DOCUMENTO', as_index=False).agg({
    'Fecha Inicio': 'max',
    'Fecha Fin': 'max'
})

# Mostrar resultado
display(df_agrupado)
```

```python
import pandas as pd

# Asegurarse que las fechas estén en formato datetime
# Convertir usando el formato correcto: MM/DD/YYYY HH:MM:SS
df_merged['Fecha Inicio_y'] = pd.to_datetime(df_merged['Fecha Inicio_y'], format='%d/%m/%Y %H:%M:%S', errors='coerce')
df_merged['Fecha Fin_y'] = pd.to_datetime(df_merged['Fecha Fin_y'], format='%d/%m/%Y %H:%M:%S', errors='coerce')

df_agrupado['Fecha Inicio'] = pd.to_datetime(df_agrupado['Fecha Inicio'], format='%d/%m/%Y %H:%M:%S', errors='coerce')
df_agrupado['Fecha Fin'] = pd.to_datetime(df_agrupado['Fecha Fin'], format='%d/%m/%Y %H:%M:%S', errors='coerce')

# Merge por COD_DOCUMENTO
df_comparado = pd.merge(df_merged, df_agrupado, on='COD_DOCUMENTO', how='left')

# Función para evaluar diferencias
def evaluar_diferencias(row):
    obs = []
    # Calcular diferencias en minutos
    dif_inicio = abs((row['Fecha Inicio_y'] - row['Fecha Inicio']).total_seconds()) / 60
    dif_fin = abs((row['Fecha Fin_y'] - row['Fecha Fin']).total_seconds()) / 60

    # Evaluar tolerancia
    if dif_inicio > 10 and dif_fin > 10:
        return f'Diferencia Hora-Inicio ({int(dif_inicio)} min) y Hora Fin ({int(dif_fin)} min)'
    elif dif_inicio > 10:
        return f'Diferencia Hora Inicio ({int(dif_inicio)} min)'
    elif dif_fin > 10:
        return f'Diferencia Hora Fin ({int(dif_fin)} min)'
    else:
        return ''

# Aplicar evaluación
df_comparado['OBSERVACION_HORARIA'] = df_comparado.apply(evaluar_diferencias, axis=1)

# Mostrar resultado
display(df_comparado)
```

```python
# Exportar a Excel
df_comparado.to_excel(r'C:\MICHAEL-CALIDAD-ENERGIA\PROYECTOS\IDENTIFICACION DE LIBRES CON FALSA GENERACION\REPORTE_COMPARADO.xlsx', index=False)
```
