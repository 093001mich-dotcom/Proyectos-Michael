```python
import pandas as pd



# 📄 Cargar el archivo principal
ruta_txt = r"C:\MICHAEL\AUTOMATIZACION DE LABORES\INSERCION DE  UBIGE\EDNA25S1_Modificado_Etapa1.txt"
ruta_xlsx = r"C:\MICHAEL\AUTOMATIZACION DE LABORES\INSERCION DE  UBIGE\ubi.xlsx"
anchos = [3, 4, 2, 10, 6, 2, 4, 4, 4, 8, 8, 8, 18, 14, 14]

# 📄 Leer el archivo y convertir todos los campos a texto desde el inicio
df = pd.read_fwf(
    ruta_txt,
    widths=anchos,
    header=None,
    dtype=str  # 👈 Forzar que todas las columnas se lean como texto
)
#Crear Fila ID
df.insert(0, "ID", [str(i) for i in range(1, len(df) + 1)])


# ✅ Asignar nombres de columnas
df.columns = ["ID",
    "CodEmpresa", "Año", "Semestre", "Suministro", "Ubigeo", "Tensión",
    "Int_NO_PROG", "Int_MANT", "Int_EXP",
    "Dur_NO_PROG", "Dur_MANT", "Dur_EXP",
    "ERS_kWh", "Comp_Punto_USD", "Comp_Ley_USD"
]

display(df)
#Etapa 2

#Eliminar filas ID seleccionadas importadas desde un excel
ruta_id = r"C:\MICHAEL\AUTOMATIZACION DE LABORES\INSERCION DE  UBIGE\eliminar suministros en CI1\Detallecabecera.xls"
df_id = pd.read_excel(ruta_id, dtype=str,sheet_name="SUMI")
#Eliminar filas con ID de la columna "fila" que están en el excel  Detallecabecera.xls
filas_eliminadas = df[df["ID"].isin(df_id["fila"].astype(str))]
df = df[~df["ID"].isin(df_id["fila"].astype(str))]
display(df)
#Mostras las filas que fueron eliminadas

display(filas_eliminadas)

# 📄 Eliminar fila ID
df = df.drop(columns=["ID"])

display(df)

ruta_salida = r"C:\MICHAEL\AUTOMATIZACION DE LABORES\INSERCION DE  UBIGE\EDNA25S1_Modificado.txt"

# 📤 Guardar en archivo .txt
with open(ruta_salida, "w", encoding="utf-8") as f:
    for _, row in df.iterrows():
        linea = "".join(str(row[col]).rjust(width)[:width] for col, width in zip(df.columns, anchos))
        f.write(linea + "\n")



print("✅ Archivo exportado correctamente en formato de ancho fijo.")
```
